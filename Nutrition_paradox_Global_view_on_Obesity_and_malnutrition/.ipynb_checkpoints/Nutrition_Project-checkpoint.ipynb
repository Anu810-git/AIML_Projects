```python
import pycountry
import numpy as np
```

```python
special_cases = {
    'GLOBAL': 'Global',
    'WB_LMI': 'Low & Middle Income',
    'WB_HI': 'High Income',
    'WB_LI': 'Low Income',
    'EMR': 'Eastern Mediterranean Region',
    'EUR': 'Europe',
    'AFR': 'Africa',
    'SEAR': 'South-East Asia Region',
    'WPR': 'Western Pacific Region',
    'AMR': 'Americas Region',
    'WB_UMI': 'Upper Middle Income'
}
```

```python
def convert_country_code(code):
    try:
        country = pycountry.countries.get(alpha_3=code)
        if country:
            return country.name
    except:
        pass
   
    return special_cases.get(code, code)  
```

```python
df_obesity1['Country'] = df_obesity1['Country'].apply(convert_country_code)
df_obesity2['Country'] = df_obesity2['Country'].apply(convert_country_code)
df_nutrition1['Country'] = df_nutrition1['Country'].apply(convert_country_code)
df_nutrition2['Country'] = df_nutrition2['Country'].apply(convert_country_code)
```

```python
df_obesity1['age_group'] = "Adult"              
df_obesity2['age_group'] = "Child/Adolescent"   
df_nutrition1['age_group'] = "Adult"            
df_nutrition2['age_group'] = "Child/Adolescent" 
```

```python
df_obesity1['CI_Width'] = df_obesity1['UpperBound'] - df_obesity1['LowerBound']
df_obesity2['CI_Width'] = df_obesity2['UpperBound'] - df_obesity2['LowerBound']
df_nutrition1['CI_Width'] = df_nutrition1['UpperBound'] - df_nutrition1['LowerBound']
df_nutrition2['CI_Width'] = df_nutrition2['UpperBound'] - df_nutrition2['LowerBound']
```

```python
df_obesity = pd.concat([df_obesity1, df_obesity2], ignore_index=True)
df_malnutrition = pd.concat([df_nutrition1, df_nutrition2], ignore_index=True)
```

```python
df_obesity
```

```python
df_malnutrition
```

```python
def classify_obesity(value):
    if value >= 30:
        return "High"
    elif value >= 25:
        return "Moderate"
    else:
        return "Low"

df_obesity['obesity_level'] = df_obesity['Mean_Estimate'].apply(classify_obesity)
```

```python
def classify_malnutrition(value):
    if value >= 20:
        return "High"
    elif value >= 10:
        return "Moderate"
    else:
        return "Low"

df_malnutrition['malnutrition_level'] = df_malnutrition['Mean_Estimate'].apply(classify_malnutrition)
```

## Exploartory Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)
```

#### 1.Shape and structure

```python
print("df_obesity shape:", df_obesity.shape)
print("df_malnutrition shape:", df_malnutrition.shape)

print("\ndf_obesity info:")
df_obesity.info()

print("\ndf_malnutrition info:")
df_malnutrition.info()

print("\ndf_obesity sample:")
display(df_obesity.head())

print("\ndf_malnutrition sample:")
display(df_malnutrition.head())
```

### 2. Missing values & unusual values

```python
print("Missing values - Obesity:\n", df_obesity.isnull().sum())
print("\nMissing values - Malnutrition:\n", df_malnutrition.isnull().sum())

# Basic stats to catch outliers/unusual values (e.g. negative CI_Width, Mean_Estimate > 100)
print("\nObesity describe:\n", df_obesity.describe())
print("\nMalnutrition describe:\n", df_malnutrition.describe())

# Sanity checks
print("\nAny negative CI_Width (obesity)?", (df_obesity['CI_Width'] < 0).any())
print("Any Mean_Estimate > 100 (obesity)?", (df_obesity['Mean_Estimate'] > 100).any())
print("Unique Year range (obesity):", df_obesity['Year'].min(), "-", df_obesity['Year'].max())
print("Unique Gender values:", df_obesity['Gender'].unique())
```

### 3. Distribution of Mean_Estimate and CI_Width

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

sns.histplot(df_obesity['Mean_Estimate'], kde=True, ax=axes[0,0], color='tomato')
axes[0,0].set_title('Obesity: Mean_Estimate Distribution')

sns.histplot(df_obesity['CI_Width'], kde=True, ax=axes[0,1], color='steelblue')
axes[0,1].set_title('Obesity: CI_Width Distribution')

sns.histplot(df_malnutrition['Mean_Estimate'], kde=True, ax=axes[1,0], color='tomato')
axes[1,0].set_title('Malnutrition: Mean_Estimate Distribution')

sns.histplot(df_malnutrition['CI_Width'], kde=True, ax=axes[1,1], color='steelblue')
axes[1,1].set_title('Malnutrition: CI_Width Distribution')

plt.tight_layout()
plt.show()
```

### 4. Trends over years (line plot)

```python
obesity_trend = df_obesity.groupby('Year')['Mean_Estimate'].mean().reset_index()
malnutrition_trend = df_malnutrition.groupby('Year')['Mean_Estimate'].mean().reset_index()

plt.figure(figsize=(12, 6))
plt.plot(obesity_trend['Year'], obesity_trend['Mean_Estimate'], marker='o', label='Obesity')
plt.plot(malnutrition_trend['Year'], malnutrition_trend['Mean_Estimate'], marker='o', label='Malnutrition')
plt.title('Average Mean_Estimate Over Years: Obesity vs Malnutrition')
plt.xlabel('Year')
plt.ylabel('Mean Estimate (%)')
plt.legend()
plt.show()
```

### 5. Top/bottom regions or countries (bar chart)

```python
top10_obesity_countries = df_obesity.groupby('Country')['Mean_Estimate'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top10_obesity_countries.values, y=top10_obesity_countries.index, palette='Reds_r')
plt.title('Top 10 Countries by Average Obesity Mean_Estimate')
plt.xlabel('Mean Estimate (%)')
plt.show()

bottom10_obesity_countries = df_obesity.groupby('Country')['Mean_Estimate'].mean().sort_values().head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=bottom10_obesity_countries.values, y=bottom10_obesity_countries.index, palette='Blues_r')
plt.title('Bottom 10 Countries by Average Obesity Mean_Estimate')
plt.xlabel('Mean Estimate (%)')
plt.show()

# By region
region_avg = df_obesity.groupby('Region')['Mean_Estimate'].mean().sort_values(ascending=False)
plt.figure(figsize=(10, 5))
sns.barplot(x=region_avg.values, y=region_avg.index, palette='viridis')
plt.title('Average Obesity Mean_Estimate by Region')
plt.show()
```

## Obesity table quries

```python
# 1. Top 5 regions with highest average obesity in 2022


select_query = """
SELECT Region, AVG(Mean_Estimate) AS avg_obesity
FROM obesity_table
WHERE Year = 2022
GROUP BY Region
ORDER BY avg_obesity DESC
LIMIT 5;
"""

df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 2. Top 5 countries with highest obesity estimates


select_query = """
SELECT Country, AVG(Mean_Estimate) AS avg_obesity
FROM obesity_table
GROUP BY Country
ORDER BY avg_obesity DESC
LIMIT 5;
"""

df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 3. Obesity trend in India over the years

select_query = """
SELECT Year, AVG(Mean_Estimate) AS avg_obesity
FROM obesity_table
WHERE Country = 'India'
GROUP BY Year
ORDER BY Year;
"""

df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 4. Average obesity by gender

select_query = """
SELECT Gender, AVG(Mean_Estimate) AS avg_obesity
FROM obesity_table
GROUP BY Gender;
"""

df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 5. Country count by obesity level category and age group

select_query = """
SELECT Obesity_level, Age_Group, COUNT(DISTINCT Country) AS country_count
FROM obesity_table
GROUP BY Obesity_level, Age_Group
ORDER BY Obesity_level, Age_Group;
"""

df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 6. Top 5 least reliable (highest avg CI_Width) 

select_query = """
SELECT Country, AVG(CI_Width) AS avg_ci_width
FROM obesity_table
GROUP BY Country
ORDER BY avg_ci_width DESC
LIMIT 5;
"""
df1 = pd.read_sql(select_query, engine)
df1
```

```python
# Top 5 most consistent (lowest avg CI_Width)

select_query = """
SELECT Country, AVG(CI_Width) AS avg_ci_width
FROM obesity_table
GROUP BY Country
ORDER BY avg_ci_width ASC
LIMIT 5;
"""
df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 7. Average obesity by age group

select_query = """
SELECT Age_Group, AVG(Mean_Estimate) AS avg_obesity
FROM obesity_table
GROUP BY Age_Group;
"""
df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 8. Top 10 countries with consistent low obesity (low average + low CI) over the years

select_query = """
SELECT Country, AVG(Mean_Estimate) AS avg_obesity, AVG(CI_Width) AS avg_ci_width
FROM obesity_table
GROUP BY Country
ORDER BY avg_obesity ASC, avg_ci_width ASC
LIMIT 10;
"""
df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 9. Countries where female obesity exceeds male by large margin (same year)

select_query = """
SELECT f.Country, f.Year,
       f.Mean_Estimate AS female_estimate,
       m.Mean_Estimate AS male_estimate,
       (f.Mean_Estimate - m.Mean_Estimate) AS difference
FROM obesity_table f
JOIN obesity_table m
  ON f.Country = m.Country AND f.Year = m.Year
WHERE f.Gender = 'Female' AND m.Gender = 'Male'
ORDER BY difference DESC
LIMIT 10;
"""
df1 = pd.read_sql(select_query, engine)
df1

```

```python
# 10. Global average obesity percentage per year

select_query = """
SELECT Year, AVG(Mean_Estimate) AS global_avg_obesity
FROM obesity_table
GROUP BY Year
ORDER BY Year;
"""
df1 = pd.read_sql(select_query, engine)
df1
```

## Combined 5 queries 

```python
# 1. Obesity vs malnutrition comparison by country (any 5 countries)

select_query = """
SELECT o.Country,
       AVG(o.Mean_Estimate) AS avg_obesity,
       AVG(m.Mean_Estimate) AS avg_malnutrition
FROM obesity_table o
JOIN malnutrition_table m
  ON o.Country = m.Country AND o.Year = m.Year
WHERE o.Country IN ('India', 'Nigeria', 'Brazil', 'United States', 'Germany')
GROUP BY o.Country;

"""
df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 2. Gender-based disparity in both obesity and malnutrition

select_query = """
SELECT o.Gender,
       AVG(o.Mean_Estimate) AS avg_obesity,
       AVG(m.Mean_Estimate) AS avg_malnutrition
FROM obesity_table o
JOIN malnutrition_table m
  ON o.Gender = m.Gender AND o.Year = m.Year AND o.Country = m.Country
GROUP BY o.Gender;

"""
df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 3. Region-wise average estimates side-by-side (Africa and America)

select_query = """
SELECT o.Region,
       AVG(o.Mean_Estimate) AS avg_obesity,
       AVG(m.Mean_Estimate) AS avg_malnutrition
FROM obesity_table o
JOIN malnutrition_table m
  ON o.Region = m.Region AND o.Year = m.Year
WHERE o.Region IN ('Africa', 'Americas')
GROUP BY o.Region;

"""
df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 4. Countries with obesity up & malnutrition down sql

select_query = """
WITH obesity_yearly AS (
    SELECT Country, Year, AVG(Mean_Estimate) AS avg_estimate
    FROM obesity_table
    GROUP BY Country, Year
),
malnutrition_yearly AS (
    SELECT Country, Year, AVG(Mean_Estimate) AS avg_estimate
    FROM malnutrition_table
    GROUP BY Country, Year
),
obesity_range AS (
    SELECT Country, MIN(Year) AS min_year, MAX(Year) AS max_year
    FROM obesity_yearly
    GROUP BY Country
),
malnutrition_range AS (
    SELECT Country, MIN(Year) AS min_year, MAX(Year) AS max_year
    FROM malnutrition_yearly
    GROUP BY Country
)
SELECT orng.Country,
       (o2.avg_estimate - o1.avg_estimate) AS obesity_change,
       (m2.avg_estimate - m1.avg_estimate) AS malnutrition_change
FROM obesity_range orng
JOIN obesity_yearly o1 ON o1.Country = orng.Country AND o1.Year = orng.min_year
JOIN obesity_yearly o2 ON o2.Country = orng.Country AND o2.Year = orng.max_year
JOIN malnutrition_range mrng ON mrng.Country = orng.Country
JOIN malnutrition_yearly m1 ON m1.Country = mrng.Country AND m1.Year = mrng.min_year
JOIN malnutrition_yearly m2 ON m2.Country = mrng.Country AND m2.Year = mrng.max_year
WHERE (o2.avg_estimate - o1.avg_estimate) > 0
  AND (m2.avg_estimate - m1.avg_estimate) < 0
ORDER BY obesity_change DESC;

"""
df1 = pd.read_sql(select_query, engine)
df1
```

```python
# 5. Age-wise trend analysis

select_query = """
SELECT o.Age_Group, o.Year, o.avg_obesity, m.avg_malnutrition
FROM (
    SELECT Age_Group, Year, AVG(Mean_Estimate) AS avg_obesity
    FROM obesity_table
    GROUP BY Age_Group, Year
) o
JOIN (
    SELECT Age_Group, Year, AVG(Mean_Estimate) AS avg_malnutrition
    FROM malnutrition_table
    GROUP BY Age_Group, Year
) m
  ON o.Age_Group = m.Age_Group AND o.Year = m.Year
ORDER BY o.Age_Group, o.Year;

"""
df1 = pd.read_sql(select_query, engine)
df1
```
